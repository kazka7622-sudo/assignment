```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("marketing_campaign.csv", sep='\t')
df.head()
```

```python
print("Shape")
df.shape
```

```python
print("Columns")
df.columns
```

```python
print("Info")
df.info()
```

```python
print("Describe")
df.describe()
```

```python
df.isnull().sum()
```

```python
plt.figure(figsize=(10,5))
sns.heatmap(df.isnull(), cbar=False)
plt.title("Missing Values")
plt.show()
```

```python
df['Income'].fillna(df['Income'].median(), inplace=True)
```

```python
df['Income'].isnull().sum()
```

```python
print("Duplicates:", df.duplicated().sum())
```

```python
print(df.dtypes)
```

```python
df['Dt_Customer'] = pd.to_datetime(df['Dt_Customer'], format='%d-%m-%Y')
```

```python
print(df['Dt_Customer'].dtypes)
```

```python
current_year = 2025
df['Age'] = current_year - df['Year_Birth']
```

```python
print(df[['Year_Birth','Age']].head())
```

```python
df['Total_Spending'] = (
      df['MntWines']
    + df['MntFruits']
    + df['MntMeatProducts']
    + df['MntFishProducts']
    + df['MntSweetProducts']
    + df['MntGoldProds']
)
```

```python
print(df[['Total_Spending']].head())
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df['Income'])
plt.title("Income Outliers")
plt.show()
```

```python
Q1 = df['Income'].quantile(0.25)
Q3 = df['Income'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df = df[(df['Income'] >= lower) &
     (df['Income'] <= upper)]
```

```python
avg_income = np.mean(df['Income'])
print("Average income: ",avg_income)
```

```python
print("Maximum income: ",np.max(df['Income']))
```

```python
print("Standard deviation: ",np.std(df['Income']))
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['Income'], bins=20)
plt.title("Income Distribution")
plt.xlabel("Income")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='Education', data=df)
plt.title("Education Distribution")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Age'], bins=20)
plt.title("Age Distribution")
plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(
    df.select_dtypes(include=np.number).corr(),
    annot=True,
    cmap='coolwarm'
)
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    x='Income',
    y='Total_Spending',
    data=df
)
plt.title("Income vs Spending")
plt.show()
```
